```python
import requests
from bs4 import BeautifulSoup
import pandas as pd 
```

```python
url = "https://webscraper.io/test-sites/e-commerce/allinone/phones"
```

```python
getresponse = requests.get(url)
print(getresponse.status_code)
```

```python
print(getresponse.text)
```

```python
if getresponse.status_code == 200:
    soup = BeautifulSoup(getresponse.text, 'html.parser')
    products =soup.find_all('div',class_='product-wrapper card-body')
    product_data = []

    for product in products:
        try:
            name = product.find('a',class_='title').get_text(strip=True)
            price = product.find('h4', class_='price').get_text(strip=True)
            review = product.find('p', class_='description').get_text(strip=True)
            product_data.append({'Name': name, 'Price': price, 'Review':review})
        except Exception as e:
            print(f"Error processing product: {e}")

    df = pd.DataFrame(product_data)
    df.to_csv(
    '/Users/ismailqamri/Desktop/tablets_data.csv',
    index=False
)

    print("table details scraped")
else:
    print("Failed",getresponse.status_code)
```
